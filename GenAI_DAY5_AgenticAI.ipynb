# AI Agent란?

  전통적인 강화학습론에서는 Agent를 “환경”을 “인식”하고 그에 따라 “행동” 하는 개체로 정의합니다.  

image.png

에이전트는 **환경** 을 **관측** 하여 **정책** 에서 **행동** 을 선택하여 실행, **상태변화에 따라 보상 혹은 패널티** 를 얻습니다. 이에 보상을 최대화하는 최적의 행동을 선택하게 됩니다.

  알림을 설정하고, 날씨를 확인하고, 좋아하는 음악을 틀어주는 핸드폰의 디지털 비서가 좋은 전통적인 Agent의 예시입니다. 주로 단순 작업 자동화에 초점을 맞추었으며, 제한된 자율성과 의사결정 능력을 가졌습니다. 이러한 Agent들은 본질적으로 사용자의 지시에 따라 정해진 범위 내에서 작동합니다.

  그러다 LLM의 발전에 맞추어 언어 모델을 탑재하면서, Agent system은 사용자와 더 자연스러운 대화가 가능해지고, 방대한 정보를 처리하면서도 다양한 환경과 상호작용할 수 있는 형태로 발전하였습니다. Agent 기술은 고객 서비스, 의료, 금융, 물류 등 다양한 산업 분야에 도입되기 시작했습니다.

**Quiz : 위 그림에서 LLM을 통해 바꿀 수 있는 단계는 몇번일까요?**

A : 1단계, 2단계, 3단계, 5단계, 그리고 엄밀하게 이야기하면 2단계와 4단계도 가능합니다!

사람이 정의한 조건 내의 관측 → 자연어 추론을 통한 다양한 조건을 고려한 관측

사람의 정의한 한계 내의 행동 → 자연어 추론을 통한 한계가 없는 행동

사람이 정의한 패널티 → 다양한 형태의 패널티, 물론 실제 시스템에 반영하려면 한 단계 가공을 거쳐야 하겠지만요!

AI Agent를 정의하는 방법은 다양하지만, 행동할 수 있는 AI (Actionable AI), 그리고 스스로 의사를 결정할 수 있는 AI (Autonomous AI) 는 모든 견해에서 확인할 수 있습니다.

#  AI Agent 의 구성 요소와 작동 방식

그렇다면, 사용자가 원하는 목표를 이루기 위한 AI Agent의 구성 요소는 어떤 것들이 있을까요?

image.png

2025년 2월 구글은 Agent white paper를 통해 Agent system 의 핵심 구성 요소를 다음과 같이 정의하였습니다. (요약글 : https://modulabs.co.kr/blog/agent-whitepapers-google)

- 모델 (Model)
    모델(Model)은 에이전트(Agent)의 의사 결정을 담당하는 핵심 요소입니다. 텍스트, 이미지 생성 등 다양한 작업을 수행할 수 있는 대규모 멀티모달 모델(LMM)을 활용합니다. ReAct, Chain-of-Thought, Tree-of-Thoughts 등의 프레임워크를 따라 복잡한 추론을 수행합니다.  

- 도구 (Tools)
    에이전트(Agent)가 외부 데이터 및 서비스와 상호작용할 수 있도록 지원하는 요소입니다. API 호출, 데이터베이스 검색, 실시간 정보 접근 등 다양한 형태로 존재하며, 에이전트(Agent)가 실제 세계와 연결되는 핵심 통로 역할을 합니다.  
    
- .오케스트레이션 레이어 (Orchestration Layer)
    오케스트레이션 레이어는 에이전트가 정보를 수집하고, 추론을 수행하며, 최종 결정을 내리는 순환적인 프로세스를 관리합니다. 이 레이어는 메모리와 상태를 유지하며, 다양한 프레임워크를 사용해 복잡한 작업을 수행합니다.

모델과 Agent의 차이점을 더 명확하게 하기 위해, 중요한 부분을 짚고 넘어가겠습니다

image.png

위의 요소로 구성된 AI agent의 동작 방식은, 요리사가 복잡한 요리를 준비하는 과정에 비유할 수 있습니다.

image.png

1단계: 정보 수집

- 요리사는 고객 주문에 맞춰 식재료를 수집하며 요리를 준비합니다.
- 에이전트는 사용자 쿼리를 입력받거나 외부 데이터를 조회합니다.

2단계: 내부 추론

- 요리사는 수집된 정보를 바탕으로 어떤 요리를 만들지 결정합니다.
- 에이전트는 모델을 활용하여 다음 행동을 결정합니다.

3단계: 행동 실행

- 요리사는 재료를 손질하고 요리를 만듭니다.
- 에이전트는 도구를 사용하여 외부 시스템과 상호작용합니다.

4단계: 피드백 및 조정

- 요리사가 요리 중간에 맛을 보고, 부족한 부분을 추가하거나 요리 순서를 변경합니다.
- 에이전트는 오케스트레이션 레이어에서 이전 결과에 기반하여 다음 행동을 조정합니다.

위의 과정을 통해, 요리사와 에이전트는 과정을 반복하며 점진적으로 목표를 달성하게 됩니다.

사용자는 이러한 에이전트의 동작을 오케스트레이션 레이어에서 메모리, 상태, 추론, 계획을 관리할 수 있습니다. 특히 추론의 경우, 기존의 LLM 혹은 RAG system보다 확장된 형태의 Multi hop 추론이 가능합니다.



**Quiz : 위에서 언급한 Multi-hop reasoning의 정의와 예시를 Single hop reasoning과 함께 찾아봅시다. 어떤 예시를 들어보시겠어요?**



싱글-홉 reasoning : 하나의 도구나 하나의 액션만으로 목표를 달성하는 방식입니다. 사용자의 요청을 받으면 필요한 도구를 단일 단계로 호출하고, 그 결과를 바로 응답으로 드리는 흐름을 가집니다. 구조가 단순하고 절차가 명확하다는 특징이 있습니다.

멀티-홉 reasoning : 더 복잡한 문제를 해결하기 위해 여러 도구를 순차적으로 또는 조건적으로 활용하는 방식입니다. 먼저 문제를 이해하고 전체적인 계획을 세운 뒤, 단계별로 도구를 호출하며 중간 결과를 분석해 다음 행동을 결정합니다. 계획과 추론, 결과 해석이 반복되는 구조라는 점이 핵심입니다.

# Agentic AI

image.png

앞서 언급한 Agent의 추론 방식이 복잡해지고, 다양한 오케스트레이션 방식과 툴을 사용하게 되면서 연구자들 사이에서는 어디까지가 에이전트이고, 어디부터가 시스템인가 보다 심도있는 논의가 시작되었습니다. 이에 AI의 대가 Andrew Ng 을 비롯한 연구자들은 다음의 경우를 Agentic AI로 재정의하기 시작하였습니다. https://www.deeplearning.ai/the-batch/issue-253/

AI 시스템이 복잡한 환경에서 사용자 지시 없이 어려운 목표를 추구할 수 있는 경우

- 자연어 지시 하에 사용자의 인터페이스나 감독 없이 자율적으로 행동할 수 있는 경우
- 웹 검색, 프로그래밍 등의 도구를 사용할 수 있거나 계획할 수 있는 경우

위에 언급한 Agentic 하다는 단어가 기존의 Agent 시스템 간의 차이를 잘 설명할 수 있는 단어가 될 것 같습니다.

image.png

그렇다면 Agentic AI 와 AI Agent 로 칭할 수 있는 것들은 무엇이 있을까요?

- 환경에 따라 주행을 조정하는 자율주행 자동차
- 의료 데이터를 분석하고, 패턴을 식별하며, 의사가 더 정보에 입각한 결정을 내릴 수 있도록 돕는 헬스케어 시스템

등 복잡하고 상황이 변하는 것은 Agentic AI 로,

- 알림을 설정하고, 날씨를 확인하고, 좋아하는 음악을 틀어주는 개인 비서 앱
- 코드 작성을 돕는 Copilot

등 정의된 명령에 의존하며 간단하고 반복적인 작업을 처리하는 것들은 AI Agent 로 구분할 수 있습니다.

# Agentic AI 시스템의 사례

그렇다면 실제 산업에서는 Agentic AI 시스템이 어떻게 활용되고 있는지, 몇가지 사례를 살펴보며 아이디어와 효과를 확인해보겠습니다.

**Morgan Stanley “Ai debrief”**

image.png

미국의 금융기업 Morgan Stanley 는 신속하고 정확한 지식 관리 시스템을 표방한 Ai debrief 를 구축하여 금융 전문가가 쉽게 고객 포트폴리오를 구축할 수 있도록 돕고 있습니다. 투자 전략, 시장 분석, 상품 정보처럼 방대한 데이터를 AI가 즉시 검색하고 분석하며, 고객별 포트폴리오와 시장 상황을 고려해 개인화된 투자 조언을 생성합니다. 이를 통해 컨설턴트의 정보 검색 시간이 50% 단축되고, 고객 응대 품질이 향상되었습니다. 더불어 복잡한 금융 상품을 설명하는 능력이 높아지면서 고객 만족도 역시 증가하는 효과가 나타났습니다.

**Klarna “Ai assistant”**

image.png

스웨덴의 핀테크 기업 Klarna는 “AI Assistant”를 도입해 고객 응대 효율을 크게 높인 것으로 알려졌습니다. 출시 첫 달 동안 약 230만 건의 대화를 처리했으며, 전체 문의의 3분의 2를 AI가 담당했습니다. 이는 약 700명 규모의 인력에 해당하는 처리량으로 평가되었습니다.

이 도입 이후 재문의 비율이 25% 감소했고, 문의 처리 시간도 평균 11분에서 2분으로 크게 단축된 것으로 보고되었습니다.

**Expedia “Romie”**

image.png

Expedia에서 여행 준비부터 진행까지 전 과정을 도와주는 AI 비서 서비스 ‘Romie’는 사용자 그룹 채팅에 직접 참여하여 조언을 제공합니다. 스마트 검색 기능을 통해 채팅 내용을 요약하고 이를 Expedia 검색 결과에 반영해 보다 정교한 추천을 제시할 수 있습니다. 일정 구성 단계에서는 이메일 정보와 호텔 주변 데이터를 활용해 맛집과 활동을 자동으로 추천하며, 날씨나 돌발 상황이 발생했을 때에는 빠르게 대안을 제시하는 방식으로 문제 대응을 지원할 수 있습니다. 실시간 업데이트 기능을 통해 일정을 자동으로 갱신할 수 있고, 공항 도착 여부 등도 즉시 확인할 수 있도록 제공할 수 있습니다. 이러한 기능 덕분에 기존 대비 약 50%의 고객이 스스로 여행 계획을 세우는 비율이 증가한 것으로 보고되었습니다.

**내부 구조도 확인할 수 있는 심층 사례도 확인해봅시다!**

**Capital one - Chat concierge**

image.png

Capital one은 미국의 대형 금융서비스 회사로, 신용카드 발급과 소비자 뱅킹, 그리고 특히 자동차 대출/리스 서비스를 운영하고 있습니다. 고객 응대에 Multi agent system으로 구축된 Chat concierge라는 서비스를 도입하였습니다. 이 서비스는 딜러들이 구독형으로 사용할 수 있는 도구로, 자동차 모델 및 기능에 대한 문의부터,보상 판매, 금융 상담까지 처리 가능합니다.

image.png

공개된 Chat Concierge 구조는 다음과 같습니다.

- 커뮤니케이터 : 다양한 형태의 고객 질의를 정련
- 플래너: 고객 질의에 응대할 내용을 구조화하여 계획
- 평가자: 플래너의 시나리오가 적합한지 검사
- 설명자: 계획을 바탕으로 응답을 생성

내부적으로는 다음과 같은 형태를 가지는 것으로 보고되었습니다.

image.png

- 툴 / 데이터 레이어 + : 재고량, 고객 응대 내역 등을 상황에 따라 참조합니다
- 오케스트레이션 레이어 + : 고객 응대 과정의 맥락 정보를 보관하고, 필요 시에 호출하여 참조합니다.
- UX 레이어 + : 고객 경험 및 피드백을 주고받습니다.

이렇게 정교하게 만들어진 시스템이지만, 겪고있는 난항이 있다고 하는데요, 이는 모든 Agentic AI system이 겪고 있는 문제와 같습니다.

- Hallucination : LLM 여럿이 협력하면서, 한 에이전트에서 발생한 Hallucination의 나비효과는 걷잡을 수 없게 됩니다. Capital one의 엔지니어들은 에이전트간의 할루시네이션을 어떻게 핸들링 할 수 있을 지 큰 고민을 하고 있다고 하네요!
- Safety : 안전한 고객 경험 제공을 위해 Jailbraking, Backdoor 공격 등 여러 케이스 에 대비책을 세운다고 합니다.

현재까지 논의했던 agentic ai 를 손쉽게 구현할 수 있는 프레임워크를 몇 개 소개해드립니다

**Hugging Face SmolAgent**

SmolAgents는 단일 에이전트가 스스로 계획·실행·수정을 반복하는 자기 개선 방식으로 작동합니다. 복잡한 제어 구조 없이 프롬프트만으로 구동되며, 외부 개입 없이 자율적으로 출력을 평가하고 개선합니다.

구조가 단순해 구현이 쉽고 자체 교정이 필요한 작업에 효과적이나, 다중 에이전트 협업에는 적합하지 않습니다.

**주요 활용 분야:** 코드 생성 및 수정 / 문서 교정 / 단계별 문제 해결

**CrewAI**

CrewAI는 역할(Role)과 태스크(Task) 기반으로 여러 에이전트가 협업하는 방식으로 작동합니다. 복잡한 워크플로우를 각 에이전트에게 효율적으로 분담하여 실행합니다.

직관적인 설계로 워크플로우를 빠르게 구축할 수 있고 협업 시너지가 높은 반면, 유연성이 비교적 낮아 커스터마이징에 한계가 있습니다.

**주요 활용 분야:** 콘텐츠 생성 파이프라인 / 시장 조사 및 보고서 작성 / 소프트웨어 개발 프로세스

**OpenAI Swarm**

OpenAI Swarm은 여러 에이전트가 협동하거나 경쟁하며 분산된 문제를 함께 해결하는 프레임워크입니다.

복잡하고 분산된 문제 해결에 효과적이며 새로운 아이디어를 도출하는 데 강점을 보이나, 에이전트 간 상호작용 제어가 어렵고 비효율적인 충돌이 발생할 수 있습니다.

**주요 활용 분야:** 주식 예측 / 신약 개발 / 최적화 문제 해결

**Microsoft AutoGen**

Microsoft AutoGen은 사람-에이전트, 에이전트-에이전트 간의 대화를 통해 문제를 해결하는 프레임워크입니다. 강력한 코드 생성 및 실행 기능을 내장하고 있어 프로그래밍 작업에 특화되어 있습니다.

직관적인 대화형 인터페이스로 사용이 쉬운 반면, 대화 흐름이 비효율적으로 흘러가거나 예측하기 어려운 상황이 발생할 수 있습니다.

**주요 활용 분야:** 데이터 분석 및 시각화 / 복잡한 시스템 자동화 / 개발 및 버그 수정

**LangGraph**

LangGraph는 순환 구조와 조건 분기를 포함한 그래프 형태의 워크플로우로 에이전트를 구성하는 프레임워크입니다. LangChain 컴포넌트와 완벽하게 통합되어 기존 LangChain 생태계를 그대로 활용할 수 있습니다.

복잡하고 동적인 의사결정 프로세스 구현에 최적화되어 있으나, 개념 난이도가 높아 단순한 작업에는 오히려 비효율적일 수 있습니다. 현업에서 많이 쓰이는 프레임워크이기도 합니다.

**주요 활용 분야:** 복잡한 의사결정 프로세스 자동화 / 반복적 수정 및 검토 시스템 / 장기 기억을 활용하는 대화형 시스템
